##🔥 Day 1 + 2 Mini Project: Student Analyzer v1

This project will help you practice:

Creating DataFrames,
Finding top students,
Calculating averages,
Filtering data,
Handling missing values,
Using apply() and lambda

```python
import pandas as pd
import numpy as np

data = {
    "name": ["Raj", "Shaam", "Priya", "Neha", "Rohan"],
    "marks": [85,45, np.nan, 72, 38]
}

df = pd.DataFrame(data)
print("Orignal Dataset:")
print(df)

# Handle Missing Data (Replace missing marks with the average marks)

avg_marks = df["marks"].mean()
df["marks"] = df["marks"].fillna(avg_marks)
print("\nAfter handling Missing Data:")
print(df)

# Show Top Students
top_student = df.loc[df["marks"].idxmax()]
print("\nTop Student:")
print(top_student)

# Show Average Marks
average = df["marks"].mean()
print("\nAverage Marks:", round(average,2))

# Filter Students Scoring More Than 50
passed_students = df[df["marks"] > 50]
print("\nStudents Scoring Above 50:")
print(passed_students)
```

### Data Visualization Project

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = {
    "name": ["raj", "Shaam", "Rohit", "Priya", "Neha"],
    "marks": [50,60,70,80,90]
}

df = pd.DataFrame(data)

# Finding-Correlation
print(df.corr(numeric_only=True))

# Outliers
new_row = {"name": "Zara", "marks": 100}
df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)
print(df)

# Scatter-Plot
plt.scatter(df["name"], df["marks"])
plt.xlabel("name")
plt.ylabel("marks")
plt.title("Student Names Vs Marks")
plt.show()

#Line-Plot
plt.plot(df["name"], df["marks"])
plt.show()

# Bar-Chart
plt.bar(df["name"], df["marks"])
plt.title("Student Vs Marks")
plt.show()

# Histogram
plt.hist(df["name"])
plt.title("Marks")
plt.show()
```

### Linear-Regression

```python
# Dataset

import pandas as pd
import matplotlib.pyplot as plt
import sys
!{sys.executable} -m pip install -q scikit-learn
from sklearn.linear_model import LinearRegression

data = {
    "hours_studied": [5,8,10,12,15],
    "marks": [50,60,70,80,90]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Split Input & Output

X = df[["hours_studied"]]
y = df["marks"]

# Train Model
model = LinearRegression()
model.fit(X,y)

# Make Prediction

pred = model.predict([[12]])
print("Predicted Marks for 12 hours of study:", pred[0])
```

```python
# Visualization

plt.scatter (X,y)  # Actual Data
plt.plot(X, model.predict(X))  # Predicted Line
plt.xlabel("Hours")
plt.ylabel("Marks")
plt.title("ML Model Fit")
plt.show()
```

```python
print("Slope (m):", model.coef_)
print("Intercept (b):", model.intercept_)
```

```python
# User-Input

hours = float(input("Enter hours studied: "))
prediction = model.predict([[hours]])
print("Expected Marks:", prediction[0])
```

### 🔥 Day 4 Mini Project

Today you'll build your first Machine Learning model using Linear Regression.

The model will learn:

More Hours Studied ➜ More Marks

```python
#Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Create Dataset

data = {
    "hours_studied": [5,8,10,12,15],
    "marks":[50,60,70,80,90]
    }

df = pd.DataFrame(data)
print(df)
```

```python
# Prepare Data
X = df[["hours_studied"]]
y = df["marks"]
 
 #Train Model
model = LinearRegression()
model.fit(X,y)
print("Model Trained Successfully!")
```

```python
# Predict-Marks

prediction = model.predict([[8]])
print("Predicted Marks: ", prediction[0])
```
