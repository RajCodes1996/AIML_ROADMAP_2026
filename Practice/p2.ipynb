### Different Operations + Handlind Missing Values

```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import FileLink,display

data ={
    "hours_studied": [5,8,10,12,15],
    "marks": [50,60,None,None,90]
}

df = pd.DataFrame(data)
df["marks"] = df["marks"].fillna(df["marks"].mean())
df.to_csv('Missing_value_handled.csv', index=False)
print("CSV Saved Successfully!")
print(df)


# Data-Visualization
plt.scatter(df["hours_studied"], df["marks"])
plt.xlabel("hours_studied")
plt.ylabel("marks")
plt.title("Hours Vs Marks")
plt.show()
```

### Sorting Data & Handling Missing Values

```python
import pandas as pd

data = {
    "name": ["Raj", "Vinod", "Pawan", "Prakash", "Messi"],
    "hours_studied":[5,8,10,12,15],
    "marks": [50,60,70,80,90]
}

df = pd.DataFrame(data)

print("\n")

df.loc[2, "marks"] = None
print("\n Replace values: ", df)

print("\n")

df["marks"] = df["marks"].fillna(df["marks"].mean())
print("\n The Missing value handled successfully!", df)

print("\n")
# Sorting Values 
df_sorted = df.sort_values(by="marks", ascending=False)
print("\n The Sorted Values Are: ", df_sorted)
```

##🔥 Day 1 + 2 Mini Project: Student Analyzer v1

This project will help you practice:

Creating DataFrames,
Finding top students,
Calculating averages,
Filtering data,
Handling missing values,
Using apply() and lambda

```python
import pandas as pd
import numpy as np

data = {
    "name": ["Raj", "Shaam", "Priya", "Neha", "Rohan"],
    "marks": [85,45, np.nan, 72, 38]
}

df = pd.DataFrame(data)
print("Orignal Dataset:")
print(df)

# Handle Missing Data (Replace missing marks with the average marks)

avg_marks = df["marks"].mean()
df["marks"] = df["marks"].fillna(avg_marks)
print("\nAfter handling Missing Data:")
print(df)

# Show Top Students
top_student = df.loc[df["marks"].idxmax()]
print("\nTop Student:")
print(top_student)

# Show Average Marks
average = df["marks"].mean()
print("\nAverage Marks:", round(average,2))

# Filter Students Scoring More Than 50
passed_students = df[df["marks"] > 50]
print("\nStudents Scoring Above 50:")
print(passed_students)
```

### Data Visualization with an (Outlier) + Type Of ML Graphs

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

data = {
    "name": ["raj", "Shaam", "Rohit", "Priya", "Neha"],
    "marks": [50,60,70,80,90]
}

df = pd.DataFrame(data)

# Finding-Correlation
print(df.corr(numeric_only=True))


# Outliers
new_row = {"name": "Zara", "marks": 100}
df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True)

print(df)


# Scatter-Plot
plt.scatter(df["name"], df["marks"])
plt.xlabel("name")
plt.ylabel("marks")
plt.title("Student Names Vs Marks")
plt.show()

#Line-Plot
plt.plot(df["name"], df["marks"])
plt.show()

# Bar-Chart
plt.bar(df["name"], df["marks"])
plt.title("Student Vs Marks")
plt.show()

# Histogram
plt.hist(df["name"])
plt.title("Marks")
plt.show()
```

### Linear~Regression 

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

data = {
    "hours_studied": [5,8,10,12,15],
    "marks": [50,60,70,80,90]
}

df = pd.DataFrame(data)
print(df)

#Implementing Regression

# Split Input-Output
X = df[["hours_studied"]]
y = df["marks"]

# Train Model
model = LinearRegression()
model.fit(X,y)

# Make Predictions
pred = model.predict([[8]])
print("\n The Prediction For 8 Hours Are: ", pred[0])
```

```python
# Visualization Regression

plt.scatter(X,y)
plt.plot(X, model.predict(X))
plt.xlabel("hours_studied")
plt.ylabel("marks")
plt.title("Hours Vs Marks Regression")
plt.show()
```

```python
# Visualize Prediction Line
plt.scatter (df["hours_studied"], df ["marks"])
plt.plot(df["hours_studied"],model.predict(X))
plt.xlabel ("hours_studied")
plt.ylabel ("marks") 
plt.title ("Hours_Studied Vs Marks")
plt.show()
```
