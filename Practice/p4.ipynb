```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
#Handling Missing Values 
data = {
    "hours_studied":[5,8,10,12,15],
    "marks":[50,60,None,None,90]
}

df = pd.DataFrame(data)
df["marks"] = df["marks"].fillna(df["marks"].mean())
print(df)

#Linear-Regression

#Split/Input-Output
X = df[["hours_studied"]]
y = df["marks"]

#Train the model
model = LinearRegression()
model.fit(X,y)

#Predicting-Values
pred = model.predict([[8]])
print("The predicted values for the following hours are: ", pred[0])

#Taking User's input
hours = float(input("Enter the number of Hours: "))
pred = model.predict(X,model.predict([[hours]]))
                     
```
