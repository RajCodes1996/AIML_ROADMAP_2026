```python
#Import Libraries
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "hours_studied": [5,8,10,12,15],
    "marks": [50,60,70,80,90]
}

df = pd.DataFrame(data)

#Basic Operations
print(df)
print("\n The Average Score is: ", df["marks"].mean())
print("\n The Maximum Marks is: ", df["marks"].max())
print("\n The Dataset Description: ", df.describe())
print("\n The correlation between hours studied and marks: ", df.corr())
print("\n The information about the dataset: ", df.info())
print("\n The First 5 rows of the dataset:", df.head())
```

```python
#Data Viusalization

plt.scatter(df["hours_studied"], df["marks"])
plt.xlabel("hours_studied")
plt.ylabel("marks")
plt.title("Hours studied Vs Marks")
plt.show()
```

```python
#Filling Missing Values
import pandas as pd
from IPython.display import FileLink, display
data = {
    "hours_studieed":[5,8,10,12,15],
    "marks": [50,60,70,None,None]
}
df = pd.DataFrame(data)
df["marks"] = df["marks"].fillna(df["marks"].mean())
df.to_csv('Student_Data.csv', index=False)
print("File Saved Successfully!")
print(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Load The Dataset

data = {
    "hours_studied": [5,8,10,12,15],
    "marks": [50,60,70,80,90]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Split Input and Output
x = df[["hours_studied"]]
y = df["marks"]

# Train Model
model = LinearRegression()
model.fit(x,y)

# Make Predictions
pred = model.predict([[10]])
print("Predicted Marks for 10 hours of study:", pred[0])
```

```python
#Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

#Create Dataset
data = {
    "hours_studied": [5,8,10,12,15],
    "marks": [50,60,70,80,90]
}

df = pd.DataFrame(data)
print(df)

# Implementing Regression

# Split Input/Output 
X = df[["hours_studied"]]
y = df ["marks"]

# Train Model
model = LinearRegression()
model.fit(X,y)
print("Model Trained Sucessfully!")
# Predict Values
pred = model.predict([[8]])
print("The Prediction for 8 hours is: ", pred[0])

# Taking User Input-

hours = float(input("Enter the number of hours"))
pred = model.predict([[hours]])
print(f"Predicted Marks = {pred[0]:.2f}")

# Data visulization: 

plt.scatter(df["hours_studied"], df["marks"])
plt.plot(X, model.predict(X))
plt.xlabel("Hours_Studied")
plt.ylabel("Marks")
plt.title("Hours Vs Marks")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression 

data = {
    "hours_studied": [5,8,10,12,15],
    "marks": [50,60,70,80,90]
}
df = pd.DataFrame(data)
print(df)

# Implementing Regression

# Split Input/Output Data
X = df[["hours_studied"]]
y = df["marks"]
print("Model Trained Successfully")

# Train Model

model = LinearRegression()
model.fit(X,y)

#  Make Predictions
pred = model.predict([[8]])
print("The Predicted Values for 8 Hours are: ", pred[0])

# Taking User's Input

hours = float(input("Enter the Number of Hours: "))
pred = model.predict([[hours]])
print(f"The Predicted Value is = {pred[0]:.2f}")

# Visualization 

plt.scatter(X,y)
plt.plot(X, model.predict(X))
plt.xlabel("Hours_Studied")
plt.ylabel("Marks")
plt.title("Hours Vs Marks")
plt.show()
```
