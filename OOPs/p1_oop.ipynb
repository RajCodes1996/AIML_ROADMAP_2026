```python
# OOP-Practice
class Car:
    def __init__(self,brand,model):
        self.brand = brand
        self.model = model
        
    def Car_Start(self):
        print("Car is started!")
               
class Bike(Car):
    def __init__(self,brand,model,price):
        self.price=price
        super().__init__(brand,model)
        
    def Bike_Start(self):
        print("Bike is Started")
        
car = Car("Audi","R8")
print(car.brand)
print(car.model)
car.Car_Start()

print("\n")

bike = Bike("Ninja","300","Rs.1,50,000")
print(bike.brand)
print(bike.model)
print(bike.price)
bike.Bike_Start()
```
