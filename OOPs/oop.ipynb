```python
# Simple_OOP_Program
class Bike:
    def __init__(self, name, model, price):
        self.name = name
        self.model = model
        self.price = price
        
    def start(self):
        print("Bike Started!")
        
Automobile = Bike("Honda", "CBR", 150000)
print(Automobile.name)
print(Automobile.model)
print(Automobile.price)
Automobile.start()
```

### OOP-INHERITANCE

```python
class Vechile:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    
    def display(self):
        print("Brand:", self.brand)
        print("Speed:", self.speed, "km/h")
        
class Bike(Vechile):
    def __init__(self, brand, speed, bike_type):
        super().__init__(brand, speed)
        self.bike_type = bike_type

    def show_bike(self):
        self.display()
        print("Bike Type:", self.bike_type)
        
class Car(Vechile):
    def __init__(self, brand, speed, fuel_type):
        super().__init__(brand,speed)
        self.fuel_type = fuel_type
        
    def show_car(self):
        self.display()
        print("Fuel Type:", self.fuel_type)
        
    # Creating - Objects
bike1 = Bike("Yamaha", 120, "Sports")
car1 = Car("Toyota", 180, "Petrol")

# Calling and Displaying Data

print("Bike Details:")
bike1.show_bike()

print("\nCar Details:")
car1.show_car()
              
        
```
